```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
df_brain = pd.read_csv("brainsize.txt", sep='\t', na_values=['NA'])
men = df_brain[df_brain['Gender'] == 'Male']
women = df_brain[df_brain['Gender'] == 'Female']
features = ['FSIQ', 'VIQ', 'PIQ', 'Weight', 'Height', 'MRI_Count']
print("Файл загружен! Строк:", len(df_brain))
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=men, x='FSIQ', y='MRI_Count', color='steelblue', label='Мужчины')
sns.scatterplot(data=women, x='FSIQ', y='MRI_Count', color='coral', label='Женщины')
plt.title('MRI_Count vs FSIQ')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(7, 6))
sns.heatmap(men[features].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Корреляции (мужчины)')
plt.show()
```

```python
plt.figure(figsize=(7, 6))
sns.heatmap(women[features].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Корреляции (женщины)')
plt.show()
```
